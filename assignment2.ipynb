```python
from dotenv import load_dotenv
load_dotenv()

from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate, FewShotChatMessagePromptTemplate
from langchain_core.callbacks import StreamingStdOutCallbackHandler

chat = ChatOpenAI(
    model="gpt-5.6-terra",
    streaming=True,
    callbacks=[StreamingStdOutCallbackHandler()]
)

examples = [
    {
        "movie": "기생충",
        "answer": """감독: 봉준호
        주요 출연진: 송강호, 이선균, 조여정, 최우식, 박소담
        예산: 1,140만 달러
        흥행 수익: 2억 5,180만 달라
        장르: 블랙 코미디, 스릴러
        시놉시스: 반지하에 사는 기택네 가족이 장남 기우의 고액 과외를 시작으로 부잣집 박 사장네에 하나둘 위장 취업하면서 예상치 못한 사건에 휘말린다.""",
    },
    {
        "movie": "인터스텔라",
        "answer": """감독: 크리스토퍼 놀란
        주요 출연진: 매튜 매커너히, 앤 헤서웨이, 제시카 차스테인, 마이클 케인
        예산: 1억 6,500만 달러
        흥행 수익: 7억 7,470만 달러
        장르: SF
        시놉시스: 삭량 부족과 모래폭풍으로 인류가 멸종 위기에 놓은 미래, 전직 우주비행사 쿠퍼가 인류가 살 수 있는 새 행성을 찾아 웜홀 너머로 탐사를 떠난다.""",
    },
    {
        "movie": "타이타닉",
        "answer": """감독: 제임스 캐머런
        주요 출연진: 레오나르도 디카프리오, 케이트 윈슬렛, 빌리 제인, 캐시 베이츠
        예산: 2억 달러
        흥행 수익: 22억 6,400만 달러
        장르: 역사, 로맨스
        시놉시스: 1912년 타이타닉호에 탄 가난한 화가 잭과 상류층 여성 로즈가 신분 차이를 넘어 사랑에 빠지지만, 배가 빙산과 충돌하며 침몰하기 시작한다."""
    }
]

example_prompt = ChatPromptTemplate.from_messages([
    ("human", "{movie}에 대해 알려 주세요."),
    ("ai", "{answer}"),
])

fewshot_prompt = FewShotChatMessagePromptTemplate(
    example_prompt=example_prompt,
    examples=examples,
)

final_prompt = ChatPromptTemplate.from_messages([
    ("system", "당신은 영화 정보 전문가입니다. 반드시 예시와 똑같은 형식으로, 한국어로만 답하세요."),
    fewshot_prompt,
    ("human", "{movie}에 대해 알려 주세요."),
])

chain = final_prompt | chat
result = chain.invoke({"movie": "올드보이"})
```
