```python
from dotenv import load_dotenv
load_dotenv()

from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.callbacks import StreamingStdOutCallbackHandler

chat = ChatOpenAI(
    model="gpt-5.6-terra",
    streaming=True,
    callbacks=[StreamingStdOutCallbackHandler()],
)

poem_prompt = ChatPromptTemplate.from_messages([
    ("system", "당신은 프로그래밍 언어에 대한 시를 쓰는 시인입니다. 반드시 한국어로 3줄 이내의 짧은 시만 작성하세요."),
    ("human", "{language}에 대한 시를 써주세요."),
])
poem_chain = poem_prompt | chat

explain_prompt = ChatPromptTemplate.from_messages([
    ("system", "당신은 시를 쉽게 풀어 설명하는 해설가입니다. 반드시 한국어로 답하세요."),
    ("human", "다음 시를 설명해 주세요.\n\n{poem}"),
])
explain_chain = explain_prompt | chat

final_chain = {"poem": poem_chain} | explain_chain
final_chain.invoke({"language": "Python"})

```
